```python
import numpy as np
import matplotlib.pyplot as plt

#Task 1: Define Activation Function
def sigmoid(x):
  return 1 / (1+np.exp(-x))

#ReLU Finction
def relu(x):
  return np.maximum(0,x)

#Tanh Function
def tanh(x):
  return np.tanh(x)
```

```python
#Task 2: Generate Outputs

# Input values
x_values=np.array([-5,-3,-1,0,1,3,5])

#Apply Activation Functions
sigmoid_outputs=sigmoid(x_values)
relu_outputs=relu(x_values)
tanh_outputs=tanh(x_values)

#Display Outputs
print("Input Values:",x_values)
print("\nSigmoid Outputs:",sigmoid_outputs)
print("\nReLU Outputs:",relu_outputs)
print("\nTanh Outputs:",tanh_outputs)
```

```python
#Task 3: Visualize and Compare

#Generate Continous input calues for plotting
x=np.linspace(-5,5,100)

#Calculate Outputs
y_sigmoid=sigmoid(x)
y_relu=relu(x)
y_tanh=tanh(x)

#Plot the Functions
plt.figure(figsize=(10,6))

plt.plot(x,y_sigmoid,label='Sigmoid')
plt.plot(x,y_relu,label='ReLU')
plt.plot(x,y_tanh,label='Tanh')

#Add Labels and Title
plt.xlabel('Input Values')
plt.ylabel('Output Values')
plt.title('Comparision of Activation Functions')

plt.grid(True)
plt.legend()

plt.show()
```
